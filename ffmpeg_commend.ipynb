```python
import os
data_list=[]
input_path = 'Z:/DATA/experiment_trackball_Optomotor'
dataset=os.listdir(input_path)
#exp_date= ['220814']
#exp_name='sweeploom'
exp_name='fictrac'
file_type='.avi'

for locust in dataset:
    locust_directory=input_path + '/' + locust
    session_list = os.listdir(locust_directory)
    for thisDate in session_list:
        experiment_directory=locust_directory + '/' + thisDate
        if os.path.isdir(experiment_directory):
            experiment_list=os.listdir(experiment_directory)
            for thisExperiment in experiment_list:
                file_directory=experiment_directory + '/' + thisExperiment
                if os.path.isdir(file_directory):
                    file_list=os.listdir(file_directory)
                    for thisVideo in file_list:
                        path_of_interest= file_directory + '/' + thisVideo
                        if thisVideo.startswith(exp_name) & thisVideo.endswith(file_type):
                            data_list.append(path_of_interest)
                        elif os.path.isdir(path_of_interest):
                            subfolder_list=os.listdir(path_of_interest)
                            for subVideo in subfolder_list:
                                if subVideo.startswith(exp_name) & subVideo.endswith(file_type):
                                    data_list.append(path_of_interest + '/' + subVideo)
```
