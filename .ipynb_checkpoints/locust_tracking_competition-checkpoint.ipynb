## Locust tracking competition
### goal: figure out a simple way to track locust direction in multi-fish behavioural setup
### input data: 1 min video x 3 
### target dataset: deeplabcut mobnet_100
### output file of the target dataset: result.csv x3 (unit: degree)
#### note: the coordinate system of the dlc output: top left corner (0,0), bottom right corner (1024,1024)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
```

```python
file_dir='C:/Users/neuroPC/Desktop/locust_competition'
file_name='test_video1DLC_mobnet_100_silentdiscust_tf28Sep02shuffle1_900000.h5'
```

```python
filename=file_dir + '/' + file_name
df=pd.read_hdf(filename)
```

##### With deeplabcut, we could identify multiple body parts in an animal. Based on my observation, it would be easier to use abdomen regions to quantify animal's orientation as animals do not really bend their body and abdomen is easier to track

##### sorting the data

```python
#iloc 21 22 means abdomen1, 27 28 means abdomen3
dx=df.iloc[:, 21]-df.iloc[:, 27]
dy=df.iloc[:, 22]-df.iloc[:, 28]
orientation_vector=np.column_stack((dx.to_numpy(),dy.to_numpy()))
```

#### plot the orientation vector

```python
starting_point=0
end_point=10
X=df.iloc[starting_point:end_point, 27]#Use abdomen3 to define the arrow location x,
Y=df.iloc[starting_point:end_point, 28]#Use abdomen3 to define the arrow location y,
U=orientation_vector[starting_point:end_point,0]#From abdomen3 to abdomen1 to define the arrow direction U,
V=orientation_vector[starting_point:end_point,1]#From abdomen3 to abdomen1 to define the arrow direction V,
C=range(starting_point,end_point)
plt.figure()
ax = plt.gca()
ax.quiver(X, Y, U, V, C, angles='xy')
ax.invert_yaxis()#reverse y axis to fit the coordinate system in dlc tracking
```

#### calulating the angle of orientation vector in a cardinal direction (degree).
https://math.stackexchange.com/questions/640411/angle-to-cardinal-direction

```python
dx=df.iloc[:, 21]-df.iloc[:, 27]
dy=-1*df.iloc[:, 22]-(-1*df.iloc[:, 28])## reverse y value to fit the data into the cardinal system
orientation_vector=np.column_stack((dx.to_numpy(),dy.to_numpy()))
```

```python
degree=[]
for i in range(orientation_vector.shape[0]):
    deg=-180*(math.atan2(orientation_vector[i,1], orientation_vector[i,0])/math.pi)
    if deg <0:
        deg=deg+360
    degree.append(deg)
```

```python
output_file=file_dir + '/result1.csv'
np.savetxt(output_file, degree, delimiter=',',fmt="%f")
```
